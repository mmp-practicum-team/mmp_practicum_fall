## Управление версиями Python с помощью Pyenv

Pyenv &ndash; кроссплатформенный инструмент для управления версиями Python.

[GitHub](https://github.com/pyenv/pyenv)


Основные возможности:
- Установка/удаление версий Python. Версии существуют независимо и изолированно друг от друга. Это, в том числе, означает, что пакеты, установленные для одной версии, не будут видны для другой
- Позволяет установить глобальную версию Python для пользователя. Эта версия будет использована по умолчанию при выполнении команды **Python**
- Позволяет для каждого проекта указать специфическую версию Python. В каталоге проекта будет автоматически использована именно указанная версия

Достоинства:
- Не зависит от Python (то есть, для установки Pyenv в системе не требуется никакая «начальная» установленная версия Python)
- Требует минимальной интеграцией со средой окружения (фактически, только добавления в переменную окружения PATH)
- Не требует прав администратора, может работать локально для отдельного пользователя

Настройка:
```bash
echo 'export PYENV_ROOT="$HOME/.pyenv"' >> ~/.bashrc
echo '[[ -d $PYENV_ROOT/bin ]] && export PATH="$PYENV_ROOT/bin:$PATH"' >> ~/.bashrc
echo 'eval "$(pyenv init - bash)"' >> ~/.bashrc
```

Установка полностью локализована и изолирована. Удаление:

```bash
rm -rf ~\.pyenv
```

Для Windows отдельный проект: https://github.com/pyenv-win/pyenv-win

Использование:
```bash
pyenv <command> [<args>]
```

Несколько полезных команд:
- `versions` — список установленных версий
- `install -l` — список доступных для установки версий (там есть anaconda, miniconda)
- `install <version>` — установка версии в систему
- `global <version>` — выбор глобальной версии Python для пользователя
- `local <version>` — выбор версии Python для проекта (каталога)

## Оператор цикла while

````python
a = 5
while a > 4:
    print('Current a: {}'.format(a))
    a -= 1
````

````python
while True:  # infinity loop
    pass  # in Python 3 the equal statement is '...'
````

````python
a = 6
while a > 2:
    a -= 1
    if a == 5: continue
    if a == 3: break
    print(a)
else:
    print('else')
````

- `pass` &ndash; оператор пустого действия
- `continue` &ndash; досрочный переход на следующую итерацию цикла

## Конструкция match/case

Добавлена в Python 3.10

````python
value = '1'
if value == '1':
    print('One')
elif value == '2':
    print('Two')
else:
    print('Many')
````

````python
value = '1'
match value:
    case '1':
        print('One')
    case '2':
        print('Two')
    case _:
        print('Many')
````

Пока разница небольшая, чуть сокращённый синтаксис, но...

````python
value = None
if not value:
    print('No')
else:
    print('Yes')
````

````python
value = None
match value:
    case '':
        print('No')
    case _:
        print('Yes')
````

Здесь уже видим разницу.

Что на самом деле происходит под капотом:

````python
value = '1'
if isinstance(value, str) and value == '1':
    print('One')
elif isinstance(value, str) and value == '2':
    print('Two')
else:
    print('Many')
````

````python
value = '1'
match value:
    case '1':
        print('One')
    case '2':
        print('Two')
    case _:
        print('Many')
````

Но на этом функции не ограничиваются. Можно записывать несколько вариантов:

````python
value = '1'
match value:
    case '1' | '2':
        print('Match')
````

Можно проверять на соответствие структурам:

````python
value = 'One Two Three Four'.split()

match value:
    case a, *b, c:
        print(b, a)
    case _:
        print('Not match')
````

````python
value = 'One Two Three'.split()
match value:
    case a, b:
        print(b, a)
    case a, b, c:
        print(c, b, a)
    case _:
        print('Not match')
````

````python
value = 'One Two Three'.split()
match value:
    case 'One', *b:
        print(*b)
    case a, b, c:
        print(c, b, a)
    case _:
        print('Not match')
````

````python
value = 'One 3 Three'.split()
match value:
    case 'One', '1' | '2', *b:
        print(*b)
    case a, b, c:
        print(c, b, a)
    case _:
        print('Not match')
````

Именовать совпавшие части:

````python
value = 'One 2 Three'.split()
match value:
    case 'One', '1' | '2' as num, *b:
        print(num, *b)
    case a, b, c:
        print(c, b, a)
    case _:
        print('Not match')
````

````python
value = 'One 3 Three Four'.split()
match value:
    case a, *b, c:
        print(a, b, c)
    case _:
        print('Not match')
````

````python
И даже сопоставлять словари:
````

## Стадии жизни объекта

- Создание объекта приводит к выделению фрагмента памяти<br><br>
- У каждого объекта есть счётчик ссылок<br><br>
- Каждое новая ссылающаяся переменная увеличивает этот счётчик на единицу<br><br>
- Ссылку можно удалить с помощью оператора `del`<br><br>
- Когда счётчик ссылок на объект становится равным 0, объект удаляется<br><br>
- Удаление и создание объектов может не приводит к изменениям памяти в некоторых случаях из-за кэширования (например 0 или None)<br><br>
- Но в общем случае память выделяется и освобождается

## Слабые ссылки

- Иногда возникает необходимость вести учёт объектов только когда они используются где-нибудь ещё<br><br>
- Но само слежение уже создаёт ссылку, из-за которой объект будет оставаться в памяти<br><br>
- Слабые ссылки дают возможность вести учёт объекта без создания ссылок на них<br><br>
- Для этого используется модуль weakref<br><br>
- Классическое приложение &ndash; кэширование объектов, пересоздание которых является затратным<br><br>
- Слабые ссылки нельзя создавать на объекты базовых типов

````python
# http://old.pynsk.ru/posts/2015/Sep/19/tainstvo-standartnoi-biblioteki-slabye-ssylki-weakref/#.XkhkS9kueV4

import weakref, gc

class A:
    def __init__(self, value):
        self.value = value
    def __repr__(self):  # magic method: information for print
        return str(self.value)

a = A(10)  # create object and strong reference
d = weakref.WeakValueDictionary()  # dict with weak references

d['k'] = a
d['k']
````

````python
del a
gc.collect()
d['k']
````

- при запуске из терминала получается вывод "KeyError: 'k'"
- видимо, в ноутбуке создаются дополнительные неявные ссылки
